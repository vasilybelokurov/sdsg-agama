```python
import numpy as np
import matplotlib.pyplot as plt
import agama
pot = agama.Potential(type='Plummer', mass=1, scaleRadius=1)
t, orb = agama.orbit(potential=pot, ic=[1, 0, 0, 0, 0.5, 0.1], time=50, trajsize=2001)
E = pot.potential(orb[:, :3]) + 0.5 * np.sum(orb[:, 3:]**2, axis=1)
assert np.ptp(E) < 1e-6 * abs(E[0]), np.ptp(E)
```

```python
fig, ax = plt.subplots(figsize=(4, 4))
ax.plot(orb[:, 0], orb[:, 1], lw=0.5)
ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_aspect('equal')
fig.savefig('smoke_orbit.png', dpi=80)
print('notebook OK')
```
